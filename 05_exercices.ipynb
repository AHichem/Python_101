## Corrections — 02 Fonctions

```python
def est_pair(n):
    return n % 2 == 0

def moyenne(liste):
    if len(liste) == 0:      # ou : if not liste:
        return None
    return sum(liste) / len(liste)

def convertir_temp(valeur, vers="F"):
    if vers == "F":
        return valeur * 9 / 5 + 32
    return (valeur - 32) * 5 / 9

print(est_pair(4), est_pair(7))
print(moyenne([]), moyenne([10, 20]))
print(convertir_temp(100), convertir_temp(212, vers="C"))
print(sorted(["banane", "kiwi", "ananas", "fraise"], key=lambda m: len(m)))   # ou key=len
```

## 🏁 Mini-projet : analyse de notes d'étudiants
Tout ce que tu as vu, dans un seul exercice. Essaie de le faire **sans regarder** la solution en dessous.

1. Génère avec numpy (`seed=1`) un DataFrame de 50 étudiants : colonnes `etudiant` (`"E01"`…`"E50"`),
   `classe` (au hasard parmi `"A"`, `"B"`, `"C"`), `maths`, `physique`, `anglais` (entiers entre 0 et 20).
2. Ajoute une colonne `moyenne` (moyenne des 3 matières, arrondie à 1 décimale).
3. Ajoute une colonne `mention` via une **fonction** + `apply` (≥16 TB, ≥14 B, ≥12 AB, ≥10 Passable, sinon Recalé).
4. Combien d'étudiants par mention ?
5. Moyenne de chaque matière par classe.
6. Le top 5 des étudiants.
7. Les étudiants qui ont au moins 15 en maths **et** moins de 10 en anglais.

```python
# --- SOLUTION ---
rng = np.random.default_rng(seed=1)
n = 50

notes = pd.DataFrame({
    "etudiant": [f"E{i:02d}" for i in range(1, n + 1)],
    "classe": rng.choice(["A", "B", "C"], size=n),
    "maths": rng.integers(0, 21, size=n),
    "physique": rng.integers(0, 21, size=n),
    "anglais": rng.integers(0, 21, size=n),
})

matieres = ["maths", "physique", "anglais"]
notes["moyenne"] = notes[matieres].mean(axis=1).round(1)   # axis=1 : moyenne sur chaque ligne

def mention(moy):
    if moy >= 16:
        return "TB"
    elif moy >= 14:
        return "B"
    elif moy >= 12:
        return "AB"
    elif moy >= 10:
        return "Passable"
    return "Recalé"

notes["mention"] = notes["moyenne"].apply(mention)
notes.head()
```

```python
notes["mention"].value_counts()
```

```python
notes.groupby("classe")[matieres].mean().round(1)
```

```python
notes.nlargest(5, "moyenne")
```

```python
notes[(notes["maths"] >= 15) & (notes["anglais"] < 10)]
```
