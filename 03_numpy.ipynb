# 03 · NumPy  ⏱️ ~25 min

**NumPy** = la bibliothèque de calcul numérique. Son objet central : le **`ndarray`** (tableau à N dimensions).

Pourquoi pas une simple liste ? Un array est :
- **beaucoup plus rapide** (code en C en dessous),
- **vectorisé** : une opération s'applique à tous les éléments d'un coup, sans boucle,
- **homogène** : tous les éléments ont le même type.

Pandas est construit **au-dessus** de NumPy : comprendre NumPy aide à comprendre Pandas.

```python
import numpy as np
```

## 1. Liste vs array : la vectorisation

```python
liste = [1, 2, 3, 4]
arr = np.array([1, 2, 3, 4])

print(liste * 2)    # liste : répète la liste !
print(arr * 2)      # array : multiplie chaque élément
print(arr + 10, arr ** 2, arr / 2)

# Avec une liste il faudrait une boucle / compréhension :
print([x * 2 for x in liste])
```

## 2. Créer des arrays

```python
print(np.array([1.5, 2, 3]))          # depuis une liste
print(np.zeros(5))                    # que des 0
print(np.ones((2, 3)))                # 2 lignes, 3 colonnes de 1
print(np.full(3, 7))                  # rempli d'une valeur
print(np.arange(0, 10, 2))            # comme range : début, fin (exclue), pas
print(np.linspace(0, 1, 5))           # 5 valeurs régulièrement espacées de 0 à 1 (inclus)
print(np.eye(3))                      # matrice identité
```

### Nombres aléatoires (très utilisé pour simuler des données)

```python
rng = np.random.default_rng(seed=42)   # générateur avec seed -> résultats reproductibles
print(rng.integers(1, 7, size=5))      # 5 entiers entre 1 et 6
print(rng.random(3))                   # 3 floats entre 0 et 1
print(rng.normal(loc=170, scale=10, size=4).round(1))   # loi normale (moyenne 170, écart-type 10)
print(rng.choice(["A", "B", "C"], size=5))
```

## 3. Attributs d'un array : forme, dimensions, type

```python
m = np.array([[1, 2, 3],
              [4, 5, 6]])

print(m)
print("shape :", m.shape)   # (lignes, colonnes)
print("ndim  :", m.ndim)    # nombre de dimensions
print("size  :", m.size)    # nombre total d'éléments
print("dtype :", m.dtype)   # type des éléments

# Changer la forme
a = np.arange(12)
print(a.reshape(3, 4))      # 3 lignes x 4 colonnes (3*4 doit = 12)
print(a.reshape(3, -1))     # -1 : "calcule la dimension tout seul"
print(m.T)                  # transposée
print(m.flatten())          # remet à plat en 1D
```

## 4. Indexation et slicing (sélection)
### En 1D : comme les listes

```python
a = np.arange(10, 20)   # [10 11 ... 19]
print(a)
print(a[0], a[-1])
print(a[2:5])
print(a[::3])
```

### En 2D : `array[ligne, colonne]`

```python
m = np.arange(1, 13).reshape(3, 4)
print(m)

print(m[0, 2])        # ligne 0, colonne 2 -> 3
print(m[1])           # toute la ligne 1
print(m[:, 0])        # toute la colonne 0  (":" = "tout")
print(m[0:2, 1:3])    # sous-matrice : lignes 0-1, colonnes 1-2
print(m[-1, -1])      # dernier élément
```

### Indexation booléenne (filtrage) — ⭐ concept clé, identique en Pandas
1. Une condition sur un array produit un **masque** de `True/False`.
2. On utilise ce masque entre crochets pour garder les `True`.

```python
notes = np.array([12, 7, 15, 9, 18, 11])

masque = notes >= 10
print(masque)
print(notes[masque])                   # garde les notes >= 10
print(notes[notes > 14])               # en une ligne

# Combiner des conditions : & (et), | (ou), ~ (non) — avec des PARENTHÈSES
print(notes[(notes >= 10) & (notes < 15)])
print(notes[(notes < 8) | (notes > 16)])

# ⚠️ Pas 'and' / 'or' avec numpy/pandas -> erreur. Toujours & | ~

# Compter / modifier via un masque
print((notes >= 10).sum(), "notes au-dessus de 10")   # True vaut 1
notes_bonus = notes.copy()
notes_bonus[notes_bonus < 10] = 10                    # remplace les notes < 10
print(notes_bonus)

# np.where(condition, si_vrai, si_faux)
print(np.where(notes >= 10, "admis", "recalé"))
```

## 5. Opérations et statistiques

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print(a + b, a * b)                   # élément par élément
print(a @ b, np.dot(a, b))            # produit scalaire / matriciel

print(np.sqrt(b), np.log(b).round(2), np.exp(a).round(2), np.abs([-1, 2, -3]))

data = np.array([[4, 8, 2],
                 [6, 1, 9]])
print("somme     :", data.sum())
print("moyenne   :", data.mean())
print("médiane   :", np.median(data))
print("écart-type:", data.std().round(2))
print("min / max :", data.min(), data.max())
print("argmax    :", data.argmax())   # position (à plat) du max
```

### Le paramètre `axis` (retrouvé partout en Pandas)
- `axis=0` : on agrège **le long des lignes** → un résultat **par colonne** (on « écrase » les lignes)
- `axis=1` : on agrège **le long des colonnes** → un résultat **par ligne**

```python
print(data)
print("somme par colonne (axis=0):", data.sum(axis=0))
print("somme par ligne   (axis=1):", data.sum(axis=1))
```

## 6. Broadcasting
NumPy « étire » automatiquement les dimensions compatibles. Ex. : soustraire la moyenne de chaque colonne.

```python
data = np.array([[1.0, 200.0],
                 [3.0, 400.0],
                 [5.0, 600.0]])
moyennes = data.mean(axis=0)          # shape (2,)
print(moyennes)
print(data - moyennes)                # (3,2) - (2,) -> appliqué à chaque ligne
```

## 7. Valeurs manquantes `np.nan` et pièges classiques

```python
v = np.array([1.0, np.nan, 3.0])
print(v.mean())          # nan : le NaN "contamine" le calcul
print(np.nanmean(v))     # version qui ignore les NaN
print(np.isnan(v))

# Piège : copie vs vue
a = np.arange(5)
vue = a[1:3]       # un slice est une VUE : il partage la mémoire avec a
vue[0] = 99
print(a)           # a est modifié aussi !
copie = a[1:3].copy()   # .copy() pour une vraie copie indépendante
```

## 8. Assembler des arrays

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
print(np.concatenate([a, b], axis=0))   # empile verticalement
print(np.vstack([a, b]))                # idem
print(np.hstack([a, a]))                # côte à côte
print(np.unique([3, 1, 3, 2, 1], return_counts=True))
```
