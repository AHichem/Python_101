## 2. Variables
Une variable = **une étiquette posée sur une valeur**. On l'écrit `nom = valeur`.

Règles de nommage : lettres, chiffres, `_`, pas d'espace, ne commence pas par un chiffre.
Convention Python : `snake_case` (ex. `prix_total`). Python est **sensible à la casse** (`age` ≠ `Age`).

```python
age = 28
prenom = "Hichem"
taille = 1.80
est_etudiant = False

print(prenom, "a", age, "ans")

age = age + 1      # on peut réassigner une variable
age += 1           # raccourci équivalent (existe aussi : -=, *=, /=)
print(age)

# Affectation multiple
x, y = 10, 20
x, y = y, x        # échange de valeurs en une ligne
print(x, y)
```

## 3. Les types de base

| Type | Exemple | Description |
|---|---|---|
| `int` | `42` | entier |
| `float` | `3.14` | nombre à virgule (avec un **point** !) |
| `str` | `"texte"` ou `'texte'` | chaîne de caractères |
| `bool` | `True` / `False` | booléen (majuscule obligatoire) |
| `NoneType` | `None` | « rien », absence de valeur |

`type(x)` donne le type d'une variable.

```python
print(type(42), type(3.14), type("salut"), type(True), type(None))

# Conversions de type (« casting »)
print(int("12") + 1)      # str -> int
print(float("3.5"))       # str -> float
print(str(100) + " €")    # int -> str (on ne peut pas additionner str + int directement)
print(int(7.9))           # tronque -> 7
print(round(7.9))         # arrondit -> 8
print(bool(0), bool(""), bool([]), bool("abc"))  # 0, "" et listes vides sont "faux"
```

## 4. Opérateurs

```python
a, b = 17, 5
print("addition      :", a + b)
print("soustraction  :", a - b)
print("multiplication:", a * b)
print("division      :", a / b)    # toujours un float
print("division ent. :", a // b)   # quotient entier
print("modulo (reste):", a % b)    # utile pour tester pair/impair
print("puissance     :", a ** 2)

# Comparaisons -> renvoient un bool
print(a > b, a == b, a != b, a <= 17)

# Logique : and, or, not
print(a > 10 and b > 10, a > 10 or b > 10, not a > 10)
```

## 5. Chaînes de caractères (`str`)

```python
phrase = "Python c'est génial"

print(len(phrase))            # longueur
print(phrase.upper())         # MAJUSCULES
print(phrase.lower())         # minuscules
print(phrase.replace("génial", "top"))
print(phrase.split(" "))      # découpe -> liste de mots
print("-".join(["a", "b", "c"]))  # l'inverse de split
print("  espaces  ".strip())  # enlève les espaces au début/à la fin
print("génial" in phrase)     # test d'appartenance

# Indexation : on commence à 0 !
print(phrase[0], phrase[-1])  # premier et dernier caractère
print(phrase[0:6])            # du caractère 0 inclus au 6 EXCLU -> "Python"
```

### Les f-strings (à utiliser tout le temps)
On met un `f` devant les guillemets et des variables entre `{}`.

```python
prenom, age, prix = "Hichem", 28, 1234.5678
print(f"{prenom} a {age} ans")
print(f"L'an prochain : {age + 1} ans")
print(f"Prix : {prix:.2f} €")          # 2 décimales
print(f"Grand nombre : {1234567:,}")   # séparateur de milliers
print(f"Pourcentage : {0.256:.1%}")    # format pourcentage
```

## 6. Les listes `list` — collection **ordonnée** et **modifiable**
Crochets `[ ]`. Peut contenir n'importe quoi (même des types mélangés).

```python
fruits = ["pomme", "banane", "cerise"]

print(fruits[0])        # premier élément
print(fruits[-1])       # dernier élément
print(len(fruits))      # nombre d'éléments

fruits.append("kiwi")          # ajoute à la fin
fruits.insert(1, "mangue")     # insère à la position 1
fruits.remove("banane")        # supprime par valeur
dernier = fruits.pop()         # retire et renvoie le dernier
fruits[0] = "poire"            # modifie un élément
print(fruits, "| retiré :", dernier)

print("cerise" in fruits)      # appartenance
```

### Slicing (découpage) : `liste[debut:fin:pas]`
`debut` est **inclus**, `fin` est **exclue**. Valable pour listes, str, tuples… et plus tard numpy/pandas.

```python
nombres = [0, 10, 20, 30, 40, 50, 60]
print(nombres[1:4])    # [10, 20, 30]
print(nombres[:3])     # les 3 premiers
print(nombres[3:])     # à partir de l'index 3
print(nombres[-2:])    # les 2 derniers
print(nombres[::2])    # un sur deux
print(nombres[::-1])   # inversée

# Fonctions utiles
notes = [12, 8, 15, 19, 10]
print(sum(notes), min(notes), max(notes), sorted(notes), sorted(notes, reverse=True))
notes.sort()           # trie la liste EN PLACE (modifie la liste)
print(notes)
```

## 7. Les tuples `tuple` — comme une liste mais **non modifiable**
Parenthèses `( )`. On les utilise pour des données fixes (coordonnées, couleur RGB, retour multiple d'une fonction…).

```python
point = (3, 5)
couleur = (255, 128, 0)

print(point[0], point[1])
print(len(couleur))

# Unpacking (déballage) : très courant !
x, y = point
print(f"x={x}, y={y}")

# point[0] = 10   # ❌ TypeError : un tuple est immuable. Dé-commente pour voir l'erreur.

# Tuple d'un seul élément : la virgule est obligatoire
singleton = (42,)
print(type(singleton), type((42)))
```

## 8. Les dictionnaires `dict` — paires **clé → valeur**
Accolades `{ }`. On accède à une valeur par sa **clé** (pas par position). Super important : un DataFrame pandas ressemble beaucoup à un dict de colonnes.

```python
personne = {"nom": "Dupont", "age": 30, "ville": "Paris"}

print(personne["nom"])                    # accès par clé
print(personne.get("email", "inconnu"))   # .get évite l'erreur si la clé n'existe pas

personne["age"] = 31            # modifier
personne["email"] = "d@mail.fr" # ajouter
del personne["ville"]           # supprimer
print(personne)

print(personne.keys())
print(personne.values())

# Parcourir un dictionnaire
for cle, valeur in personne.items():
    print(f"{cle} -> {valeur}")
```

## 9. Les ensembles `set` — **sans doublons**, sans ordre
Pratique pour dédoublonner ou comparer des groupes.

```python
villes = ["Paris", "Lyon", "Paris", "Nice", "Lyon"]
uniques = set(villes)
print(uniques)

a = {1, 2, 3, 4}
b = {3, 4, 5}
print(a & b)   # intersection
print(a | b)   # union
print(a - b)   # différence
```

### 📌 Récap des structures
| Structure | Syntaxe | Ordonnée | Modifiable | Doublons | Usage typique |
|---|---|---|---|---|---|
| `list` | `[1, 2]` | ✅ | ✅ | ✅ | collection d'éléments |
| `tuple` | `(1, 2)` | ✅ | ❌ | ✅ | données fixes |
| `dict` | `{"a": 1}` | ✅ (ordre d'insertion) | ✅ | clés uniques | associer clé → valeur |
| `set` | `{1, 2}` | ❌ | ✅ | ❌ | unicité, opérations ensemblistes |

## 10. Conditions : `if / elif / else`
⚠️ **L'indentation (4 espaces) définit les blocs** en Python. Pas d'accolades. Et `:` à la fin de la ligne.

```python
note = 14

if note >= 16:
    mention = "Très bien"
elif note >= 14:
    mention = "Bien"
elif note >= 10:
    mention = "Passable"
else:
    mention = "Recalé"

print(mention)

# Version courte (opérateur ternaire)
statut = "admis" if note >= 10 else "refusé"
print(statut)
```

## 11. Boucles
### `for` : parcourir une séquence

```python
for fruit in ["pomme", "banane", "cerise"]:
    print(fruit)

# range(debut, fin, pas) : fin exclue
for i in range(5):          # 0, 1, 2, 3, 4
    print(i, end=" ")
print()

for i in range(2, 11, 2):   # 2, 4, 6, 8, 10
    print(i, end=" ")
print()

# enumerate : récupérer l'index ET la valeur
for i, fruit in enumerate(["pomme", "banane"]):
    print(i, fruit)

# zip : parcourir deux listes en parallèle
noms = ["Alice", "Bob"]
ages = [25, 32]
for nom, age in zip(noms, ages):
    print(f"{nom} : {age} ans")
```

### `while` : tant qu'une condition est vraie + `break` / `continue`

```python
compteur = 0
while compteur < 3:
    print("compteur =", compteur)
    compteur += 1    # sans ça : boucle infinie !

for n in range(10):
    if n == 2:
        continue     # passe directement à l'itération suivante
    if n == 5:
        break        # sort de la boucle
    print(n, end=" ")
```

## 12. Les compréhensions de liste (très « pythonique »)
`[expression for element in sequence if condition]` : créer une liste en une ligne.

```python
carres = [x ** 2 for x in range(6)]
print(carres)

pairs = [x for x in range(20) if x % 2 == 0]
print(pairs)

mots = ["python", "pandas", "numpy"]
print([m.upper() for m in mots])

# Équivalent en boucle classique :
carres_bis = []
for x in range(6):
    carres_bis.append(x ** 2)
print(carres_bis == carres)

# Compréhension de dictionnaire
longueurs = {m: len(m) for m in mots}
print(longueurs)
```
