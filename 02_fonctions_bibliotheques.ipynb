## 1. Définir une fonction
Une fonction = un bloc de code réutilisable. Mot-clé `def`, des **paramètres** entre parenthèses, et `return` pour renvoyer un résultat.

````python
def saluer(prenom):
    """Renvoie un message de bienvenue.  <- ceci est une 'docstring' (documentation)"""
    return f"Bonjour {prenom} !"

message = saluer("Hichem")
print(message)

# ⚠️ print ≠ return : print AFFICHE, return RENVOIE une valeur qu'on peut réutiliser.
def sans_return(x):
    print(x * 2)

resultat = sans_return(5)
print("resultat =", resultat)   # None, car pas de return
````

## 2. Paramètres : valeurs par défaut, arguments nommés, retours multiples

````python
def prix_ttc(prix_ht, tva=0.20):          # tva a une valeur par défaut
    return prix_ht * (1 + tva)

print(prix_ttc(100))              # utilise tva=0.20
print(prix_ttc(100, 0.055))       # argument positionnel
print(prix_ttc(tva=0.10, prix_ht=50))   # arguments nommés : l'ordre n'importe plus

def stats(liste):
    return min(liste), max(liste), sum(liste) / len(liste)   # renvoie un tuple

mini, maxi, moyenne = stats([4, 8, 15, 16, 23, 42])   # unpacking du tuple
print(mini, maxi, round(moyenne, 2))
````

## 3. `*args` et `**kwargs` (pour info — tu les verras dans la doc des bibliothèques)

````python
def somme(*args):            # nombre variable d'arguments -> tuple
    return sum(args)

def profil(**kwargs):        # nombre variable d'arguments nommés -> dict
    return kwargs

print(somme(1, 2, 3, 4))
print(profil(nom="Alice", age=30))
````

## 4. Portée des variables (scope)
Une variable créée **dans** une fonction n'existe que dans cette fonction.

````python
def calcul():
    interne = 10
    return interne * 2

print(calcul())
# print(interne)   # ❌ NameError : 'interne' n'existe pas en dehors de la fonction
````

## 5. Fonctions `lambda` (fonctions anonymes d'une ligne)
Très utilisées avec `sorted`, `map`, `filter` et surtout **pandas `.apply()`**.

````python
double = lambda x: x * 2
print(double(7))

personnes = [("Alice", 25), ("Bob", 19), ("Chloé", 32)]
print(sorted(personnes, key=lambda p: p[1]))            # tri par âge
print(list(map(lambda x: x ** 2, [1, 2, 3])))           # applique à chaque élément
print(list(filter(lambda x: x > 1, [0, 1, 2, 3])))      # garde si condition vraie
````

- **les bibliothèques externes** : à installer avec **pip** dans un terminal :

### Les 4 façons d'importer

````python
import math                      # 1. importer tout le module
print(math.sqrt(16), math.pi)

import numpy as np               # 2. avec un alias (convention universelle : np, pd, plt)
import pandas as pd

from math import floor, ceil     # 3. importer seulement certaines fonctions
print(floor(3.7), ceil(3.2))

# from math import *             # 4. tout importer sans préfixe -> ❌ à éviter (conflits de noms)

print(np.__version__, pd.__version__)   # vérifier la version installée
````

## 7. Gérer les erreurs : `try / except`
Lire une erreur : **la dernière ligne** donne le type et la cause. Les plus courantes :
`SyntaxError`, `NameError` (variable inconnue), `TypeError` (mauvais type), `ValueError` (mauvaise valeur),
`IndexError` (index hors liste), `KeyError` (clé absente du dict / colonne absente du DataFrame), `ZeroDivisionError`.

````python
def division_sure(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("⚠️ Division par zéro !")
        return None

print(division_sure(10, 2))
print(division_sure(10, 0))

saisie = "abc"
try:
    nombre = int(saisie)
except ValueError as e:
    print("Conversion impossible :", e)
finally:
    print("Ce bloc s'exécute toujours")
````

## 8. (Bonus) Les classes en 30 secondes
Tu n'en as pas besoin pour débuter, mais **tout est objet** en Python : un DataFrame est un objet de la classe `DataFrame`,
et `df.head()` est une **méthode** (fonction attachée à l'objet). Voilà à quoi ça ressemble :

````python
class CompteBancaire:
    def __init__(self, titulaire, solde=0):   # constructeur
        self.titulaire = titulaire            # attributs
        self.solde = solde

    def deposer(self, montant):               # méthode
        self.solde += montant
        return self.solde

compte = CompteBancaire("Hichem", 100)
compte.deposer(50)
print(compte.titulaire, compte.solde)
````
